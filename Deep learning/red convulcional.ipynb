```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# ========================================
# 1. CARGA Y PREPARACIÓN DE DATOS
# ========================================

# Cargar dataset CIFAR-10
(x_train, y_train), (x_test, y_test) = keras.datasets.cifar10.load_data()

# Normalizar valores de píxeles entre 0 y 1
x_train, x_test = x_train / 255.0, x_test / 255.0

# Convertir labels a categorical (one-hot encoding)
num_classes = 10
y_train_categorical = keras.utils.to_categorical(y_train, num_classes)
y_test_categorical = keras.utils.xto_categorical(y_test, num_classes)

print("Forma de los datos:")
print("x_train shape:", x_train.shape)
print("y_train shape:", y_train.shape)
print("x_test shape:", x_test.shape)
print("y_test shape:", y_test.shape)
print(f"{x_train.shape[0]} muestras de entrenamiento")
print(f"{x_test.shape[0]} muestras de prueba")

# ========================================
# 2. DEFINICIÓN DE LA ARQUITECTURA
# ========================================

model = keras.Sequential([
    keras.layers.Input(shape=(32, 32, 3)),
    keras.layers.Conv2D(32, (3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.2),
    keras.layers.Conv2D(32, (3, 3), activation='relu'),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Flatten(),
    keras.layers.Dense(512, activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(10, activation='softmax')
])

# Mostrar resumen del modelo
model.summary()

# ========================================
# 3. COMPILACIÓN DEL MODELO
# ========================================

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# ========================================
# 4. CONFIGURACIÓN DE CALLBACKS
# ========================================

# Early Stopping para evitar overfitting
early_stopping = keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

# Reducir learning rate cuando el loss se estanque
reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=5,
    min_lr=0.0001,
    verbose=1
)

# Model checkpoint para guardar el mejor modelo
model_checkpoint = keras.callbacks.ModelCheckpoint(
    'best_cifar10_model.h5',
    monitor='val_accuracy',
    save_best_only=True,
    verbose=1
)

callbacks = [early_stopping, reduce_lr, model_checkpoint]

# ========================================
# 5. ENTRENAMIENTO DEL MODELO
# ========================================

print("\n🚀 Iniciando entrenamiento...")

# Parámetros de entrenamiento
EPOCHS = 15
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.2

# Entrenar el modelo
history = model.fit(
    x_train, y_train_categorical,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    validation_split=VALIDATION_SPLIT,
    callbacks=callbacks,
    verbose=1
)

print("\n✅ Entrenamiento completado!")

# ========================================
# 6. EVALUACIÓN DEL MODELO
# ========================================

# Evaluar en conjunto de prueba
test_loss, test_accuracy = model.evaluate(x_test, y_test_categorical, verbose=0)
print(f"\n📊 Resultados en conjunto de prueba:")
print(f"Loss: {test_loss:.4f}")
print(f"Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")

# ========================================
# 7. VISUALIZACIÓN DE RESULTADOS
# ========================================

def plot_training_history(history):
    """Visualizar las curvas de entrenamiento"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Plot accuracy
    ax1.plot(history.history['accuracy'], label='Training Accuracy', color='blue')
    ax1.plot(history.history['val_accuracy'], label='Validation Accuracy', color='red')
    ax1.set_title('Model Accuracy')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Accuracy')
    ax1.legend()
    ax1.grid(True)
    
    # Plot loss
    ax2.plot(history.history['loss'], label='Training Loss', color='blue')
    ax2.plot(history.history['val_loss'], label='Validation Loss', color='red')
    ax2.set_title('Model Loss')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Loss')
    ax2.legend()
    ax2.grid(True)
    
    plt.tight_layout()
    plt.show()

# Mostrar gráficos de entrenamiento
plot_training_history(history)

# ========================================
# 8. MATRIZ DE CONFUSIÓN Y REPORTE
# ========================================

# Hacer predicciones
y_pred = model.predict(x_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = y_test.flatten()

# Nombres de las clases
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

# Matriz de confusión
plt.figure(figsize=(10, 8))
cm = confusion_matrix(y_true, y_pred_classes)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Matriz de Confusión')
plt.ylabel('Valores Reales')
plt.xlabel('Predicciones')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Reporte de clasificación
print("\n📈 Reporte de Clasificación:")
print(classification_report(y_true, y_pred_classes, target_names=class_names))

# ========================================
# 9. VISUALIZACIÓN DE PREDICCIONES
# ========================================

def plot_predictions(model, x_test, y_test, class_names, num_images=12):
    """Visualizar algunas predicciones del modelo"""
    predictions = model.predict(x_test[:num_images])
    predicted_classes = np.argmax(predictions, axis=1)
    
    plt.figure(figsize=(15, 8))
    for i in range(num_images):
        plt.subplot(3, 4, i + 1)
        plt.imshow(x_test[i])
        plt.title(f'Real: {class_names[y_test[i][0]]}\n'
                 f'Pred: {class_names[predicted_classes[i]]}\n'
                 f'Conf: {np.max(predictions[i]):.2f}')
        plt.axis('off')
    
    plt.tight_layout()
    plt.show()

# Mostrar algunas predicciones
plot_predictions(model, x_test, y_test, class_names)

# ========================================
# 10. GUARDAR MODELO FINAL
# ========================================

# Guardar el modelo completo
model.save('cifar10_cnn_final.h5')
print("\n💾 Modelo guardado como 'cifar10_cnn_final.h5'")

# Guardar solo los pesos
model.save_weights('cifar10.weights.h5')
print("💾 Pesos guardados como 'cifar10.weights.h5'")

print(f"\n🎯 Resumen final:")
print(f"   • Accuracy en test: {test_accuracy*100:.2f}%")
print(f"   • Total de épocas entrenadas: {len(history.history['loss'])}")
print(f"   • Mejor val_accuracy: {max(history.history['val_accuracy'])*100:.2f}%")
```
