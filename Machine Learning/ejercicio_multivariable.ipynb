# Regresión lineal multivariable

## Importando datos

```python
import pandas as pd
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/housing/housing.data',
            header=None, sep='\s+')
df.columns = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']
df.head()
```

## Creando modelo con sklearn

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

X = df[['RM', 'INDUS']].values
y = df['MEDV'].values.reshape(-1, 1)
sc_x = StandardScaler()
sc_y = StandardScaler()
X_std = sc_x.fit_transform(X)
y_std = sc_y.fit_transform(y)

slr = LinearRegression()
slr.fit(X_std, y_std)
```

## Análisis de resultados

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np 
x1_range = np.arange(df['RM'].min(),df['RM'].max())
x2_range = np.arange(df['INDUS'].min(),df['INDUS'].max())

X1, X2 = np.meshgrid(x1_range,x2_range)

plano = pd.DataFrame({'RM':X1.ravel(), 'INDUS':X2.ravel()})
pred = slr.predict(sc_x.transform(plano)).reshape(X1.shape)
pred = sc_y.inverse_transform(pred)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X1, X2, pred, alpha=0.4, cmap='viridis')

ax.scatter3D(df['RM'], df['INDUS'], df['MEDV'], c='red', marker='.')
ax.set_xlabel('RM (Number of rooms)')
ax.set_ylabel('INDUS (Industry)')
ax.set_zlabel('MEDV (Price)')
ax.view_init(elev=10, azim=5)
plt.show()
```

## Generando predicción con el modelo

```python
# Create sample data with both features (RM and INDUS)
sample = np.array([[5.0, 2.31]])  # 5.0 rooms and 2.31 industry level (using first row's INDUS value as example)
sample_std = sc_x.transform(sample)
price_std = slr.predict(sample_std)
print('El precio en miles es de %.3f'%
      sc_y.inverse_transform(price_std))
```
