```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

texts = tf.constant([
    "I love the movie! Super Good.",
    "very bad movie, I hate it.",
    "It's okay, but I expected more."
                     
])
labels = tf.constant([1, 0, 1])
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

vectorizer = keras.layers.TextVectorization(
    output_mode="int", # significa que 
    standardize="lower_and_strip_punctuation", #significa standardize 
    split="whitespace",
    output_sequence_length=10,
    vocabulary_size=10000,
)

vectorizer.adapt(texts)
```

```python
vectorizer.get_vocabulary()
```

```python
doc = vectorizer('i love the movie')
doc.numpy()
```

```python
model = keras.Sequential([
    vectorizer,
    keras.layers.Embedding(vectorizer.vocabulary_size(),64),
    keras.layers.GlobalAveragePooling1D(),
    keras.layers.Dense(64,activation='relu'),
    keras.layers.Dense(1,activation='sigmoid')
])
```
